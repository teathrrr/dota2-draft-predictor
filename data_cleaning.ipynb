```python
import pandas as pd
```

```python
df = pd.read_csv('dota_matches_raw.csv')

df.info()
print(df.isnull().sum())
print(df.duplicated().sum())
```

First of all, we want to analyze just all pick ranked games, as they are the most dotaish

```python
df = df[df['match_gameMode'] == 'ALL_PICK_RANKED']
```

```python
print(df.isnull().sum())
```

```python
df[df['role'].isnull()]
```

```python
df['isRandom'].fillna(False)
```

```python
print(df.isnull().sum())
```

```python
pos_map = {
    ('CORE', 'SAFE_LANE'): 'POSITION_1',
    ('CORE', 'MID_LANE'): 'POSITION_2',
    ('CORE', 'OFF_LANE'): 'POSITION_3',
    ('LIGHT_SUPPORT', 'OFF_LANE'): 'POSITION_4',
    ('HARD_SUPPORT', 'SAFE_LANE'): 'POSITION_5',
}

mask = df['position'].isna() & df['role'].notna()
df.loc[mask, 'position'] = [pos_map.get(k) for k in zip(df.loc[mask, 'role'], df.loc[mask, 'lane'])]
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=['role'])
```

```python
print(df.isnull().sum())
```

```python
df['dotaPlusHeroXp'] = df['dotaPlusHeroXp'].fillna(
    df.groupby('heroId')['dotaPlusHeroXp'].transform('median').fillna(df['dotaPlusHeroXp'].median()))
```

```python
print(df.isnull().sum())
```

```python
df['isRandom'] = df['isRandom'].eq(True)
```

```python
print(df.isnull().sum())
```

```python
df = df[df['match_durationSeconds'] > 1800]
```

```python
df = df.drop(columns=['behavior', 'role', 'lane', 'match_durationSeconds'])
```

```python
print(df.isnull().sum())
```

```python
df.to_csv('clean_data.csv', index=False)
```
